## Imports

```python
import pandas as pd
import json
```

## Reading files

```python
with open('advertising_information.json', 'r', encoding='utf-8') as file:
    advertising_info = json.load(file)

advertising_info
```

```python
prices = pd.DataFrame(advertising_info)

prices
```

```python
# Reading main tables
search = pd.read_csv('search.csv')
video = pd.read_csv('video.csv')
native = pd.read_csv('native.csv')
other = pd.read_csv('other.csv')

banner = pd.read_excel('banner.xlsx')
rich = pd.read_excel('rich.xlsx')
```

```python
# setting correct Index each to tables
columns = [
    'index',
    'user_id',
    'funnel_stage',
    'timestamp',
    'profit',
    'os'
]
organic_traffic = pd.read_csv(
    'organic_traffic.csv',
    header=None,
    names=columns)

partner_network = pd.read_csv(
    'partner_network.csv',
    header=None,
    names=columns
)
```

```python
# Checking result
print('search:', search.columns.tolist())
print('video:', video.columns.tolist())
print('native:', native.columns.tolist())
print('organic:', organic_traffic.columns.tolist())
print('other:', other.columns.tolist())
print('partner:', partner_network.columns.tolist())
print('banner:', banner.columns.tolist())
print('rich:', rich.columns.tolist())
```

```python
organic_traffic = organic_traffic.drop(columns='index')
partner_network = partner_network.drop(columns='index')
```

```python
# Checking result
organic_traffic['ad_channel'] = 'organic_traffic'
partner_network['ad_channel'] = 'partner_network'
print(organic_traffic.columns.tolist())
print(partner_network.columns.tolist())
```

```python
# Dropping Unnamed column
search = search.drop(columns='Unnamed: 0')
video = video.drop(columns='Unnamed: 0')
native = native.drop(columns='Unnamed: 0')
other = other.drop(columns='Unnamed: 0')
```

```python
# Checking result
print(search.columns.tolist())
print(video.columns.tolist())
print(native.columns.tolist())
print(other.columns.tolist())
```

```python
# Adding the column ad_channel
search['ad_channel'] = 'search'
video['ad_channel'] = 'video'
native['ad_channel'] = 'native'
organic_traffic['ad_channel'] = 'organic_traffic'
other['ad_channel'] = 'other'
partner_network['ad_channel'] = 'partner_network'
banner['ad_channel'] = 'banner'
rich['ad_channel'] = 'rich'
```

```python
search.head() # check
```

```python
# Concatenate
df = pd.concat(
    [
        search,
        banner,
        video,
        native,
        partner_network,
        rich,
        organic_traffic,
        other
    ],
    ignore_index=True
)
```

```python
# checking whole DataFrame
df
```

```python
# merging price column
df = df.merge(
    prices,
    on='ad_channel',
    how='left'
)
```

```python
# deleting duplicates if created
df[['ad_channel', 'price']].drop_duplicates()
```

```python
# setting the same register
df['funnel_stage'] = df['funnel_stage'].str.lower()
df
```

```python
# group to see profit
df.groupby('funnel_stage')['profit'].count()
```

```python
# Setting the correct column presentation as it was mentioned to me.
df = df[
    [
        'user_id',
        'ad_channel',
        'funnel_stage',
        'price',
        'profit',
        'os',
        'timestamp'
    ]
]
```

```python
# Sorting by timestamps
df = df.sort_values(
    by='timestamp'
).reset_index(drop=True)
df.head(30)
```

```python
# control
df.info()
```

```python
# saving the result to new file
df.to_csv('advertising_funnel.csv', index = False)
```
